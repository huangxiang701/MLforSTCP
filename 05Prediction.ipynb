```python
df_Polyolefins=pd.read_csv(r'.\Datasets\Reaction\FF_polyester.csv',index_col=0)
df_Polyolefins
```

```python
df2=df_Polyolefins.dropna(subset=['mass_H'])
df2
```

```python
finger=[]
sigma=80
for k in range(len(df_Polyolefins)):
    desc=np.zeros(sigma).reshape(1,-1)
    data=df_Polyolefins.iloc[k].values
    for i in range(10):
        desc[0][i]=data[i]
    desc1=data[10:].reshape(-1,sigma)
    desc2=np.zeros(sigma*(sigma-10)).reshape(-1,sigma)
    finger.append(np.vstack([desc,desc1,desc2]))
finger=np.array(finger)
```

```python
finger_shape=finger.shape
finger1=finger.reshape(finger_shape[0],finger_shape[1],finger_shape[2],1)
```

```python
TC_CNN=CNN.predict(finger1)
```

```python
TC_FFNN=FFNN.predict(df2)
```
